```python
data = {
    "prompt": "你好",
    "completion": "你好",
}
```

```python
import os
model_id = os.environ.get("STARTER_MODEL", "Qwen/Qwen2.5-0.5B-Instruct")
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
prompt_text = tokenizer.apply_chat_template(
    [{"role": "user", "content": data["prompt"]}],
    tokenize=False,
    add_generation_prompt=True,
)
print(prompt_text)
```

```python
prompt_ids = tokenizer(prompt_text, add_special_tokens=False)["input_ids"]
print("prompt_ids:", prompt_ids)
```

```python
full_text = prompt_text + data["completion"] + tokenizer.eos_token
full_ids = tokenizer(full_text, add_special_tokens=False)["input_ids"]
print("full_ids:", full_ids)
```

```python
max_length = 36
encoded = tokenizer(
        full_text,
        # full_text 已含聊天模板标记与 eos 标记，不再自动追加。
        add_special_tokens=False,
        # 超过 max_length 时，截掉末尾多出的 token。
        truncation=True,
        # 本例为方便说明假设 max_length=12；正常训练默认是 128。
        max_length=max_length,
        # 不足 max_length 时，在右侧补 PAD，便于组成 batch。
        padding="max_length",
    )
full_ids = encoded["input_ids"]
full_mask = encoded["attention_mask"]
print("full_ids:", full_ids)
print("full_mask:", full_mask)
```

```python
print("EOS:", tokenizer.eos_token, ",eos_token_id:", tokenizer.eos_token_id)
print("PAD:", tokenizer.pad_token, ",pad_token_id:", tokenizer.pad_token_id)
```
